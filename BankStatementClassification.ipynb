```python
import numpy as np

#Clean formatting issues before conversion
# Remove commas, currency symbols (₦, $, etc.), and spaces
df["WITHDRAWALS"] = df["WITHDRAWALS"].astype(str).str.replace(",", "", regex=False)
df["WITHDRAWALS"] = df["WITHDRAWALS"].str.replace("₦", "", regex=False).str.strip()
df["DEPOSITS"] = df["DEPOSITS"].astype(str).str.replace(",", "", regex=False)
df["DEPOSITS"] = df["DEPOSITS"].str.replace("₦", "", regex=False).str.strip()
```

```python
# Convert DEBIT and CREDIT to numeric (handles text like "0.00" or blanks)
df["WITHDRAWALS"] = pd.to_numeric(df["WITHDRAWALS"], errors="coerce")
df["DEPOSITS"] = pd.to_numeric(df["DEPOSITS"], errors="coerce")
```

```python
# Replace 0.00 with NaN
df["WITHDRAWALS"] = df["WITHDRAWALS"].replace(0.00, np.nan)
df["DEPOSITS"] = df["DEPOSITS"].replace(0.00, np.nan)
```

```python
from ast import keyword
def classify_transaction(DESCRIPTION, DEPOSITS, WITHDRAWALS):
    desc = str(DESCRIPTION).upper()

 # --- CREDIT SIDE ---
    if pd.notna(DEPOSITS) and DEPOSITS > 0:
        if any(keyword in desc for keyword in ["LEKKI SEAPORT", "LEKKI PORT"
        ]):
            return "Lekki Port C"
        elif "FINCRA TECHNOLOGIES LIMITED" in desc:
            return "Fincra Technologies Limited"
        elif "DEWLL INERNATIONAL" in desc or  "DEWLL INTERNATIONAL LOGISTICS LTD" in desc:
            return "DEWLL INTERNATIONAL C"
        elif "ADVANCE PAYMENT FOR CUSTOMS CLEARANCE" in desc or "XCMG" in desc:
            return "Advance Payment XCMG"
        else:
            return "Other Credit"

    # --- DEBIT SIDE ---
    if pd.notna(WITHDRAWALS) and WITHDRAWALS < 0:
        if any(keyword in desc for keyword in [
            "BANK_CHARGES", "BANK CHARGES",  "PP_FEE", "FGN STAMP DUTY", "TRSF LEVY", "SMS ALERT", "TRANSFER LEVY",
             "UNDEBITED FEES", "ONLINE PAYMENT FEE"," + VAT", "VAT ON COMMISSION","CHARGE"
        ]):
            return "Charges"
        elif "LEKKI SEAPORT" in desc or "LEKKI PORT" in desc:
            return "Lekki Seaport"
        elif "WEST MERCEDONIA NIG LIMITED" in desc:
            return "West Mercedonia"
        elif "11PLC WAREHOUSE ROAD APAPA" in desc:
            return "11 PLC"
        elif " SIFEM GLOBAL RESOURCES" in desc or "SIFEM GLOBAL RESOURCES - SGR"in desc or "SIFEM GLOBAL RESOURCES - SGR 29" in desc or "SIFEM GLOBAL RESOURCES" in desc:
            return "SIFEM Global Services"
        elif "ABAHAM ENERGY LIMITED" in desc:
            return "Abaham Energy Limited"
        elif "CARRY TRANSPORTATION NIG LTD" in desc:
            return "Carry Transportation Nig LTD"
        elif "DEWLL INTERNATIONAL LOGISTICS LTD" in desc:
            return "Dewll International"
        elif "AFRICA GLO LOG  TRAS" in desc:
            return "Africa Glo Trans"
        elif "SHAFA LIMITED" in desc:
            return "Shafa LTD"
        elif "ROLEZ" in desc:
            return "Rolez"
        elif "OWOLE SALARY" in desc or "OWOLE FLORENCE" in desc:
            return "Owole"
        elif "NATIONAL SINGLE WINDOW" in desc:
            return "Single Window"
        elif "COSCO SHIPPING" in desc or "COSCO SHIPPING LINES (NIGERIA) CO" in desc or "COSCO SHIPPING LINES  NIGERIA  CO." in desc:
            return "Cosco Shipping"
        elif "APM TERMINALS" in desc or "APMT" in desc:
            return "APMT"
        elif "INTB/S E S LTD" in desc:
            return "NTB/SES LTD"
        elif "DANCO PETROLEUM" in desc or "DANCO PETROLEUM LIMITED" in desc or "DANCO PETROLEUM LIMITED" in desc or "DANCO PET" in desc:
            return "Danco Petroleum"
        elif "CARTRIDGE" in desc:
            return "Cartridge"
        elif "OSONOCHI MARKETING" in desc:
            return "Osonoch Marketing"
        elif "BRAVE WAVES" in desc:
            return "Brave Waves"
        elif "COBITA INVEST" in desc:
            return "Cobita"
        elif "QUANLEE OVERSEAS WAREHOUSE" in desc:
            return "QUANLEE OVERSEAS WAREHOUSE"
        elif "DAVID OBATARE" in desc:
            return "David Obatare"
        elif "TRANSFER" in desc or "TRF TO" in desc or "TFR TO" in desc:
            return "Transfer"
        elif "MAINTENANCE FEE" in desc or "ACCOUNT MAINTENNACE" in desc:
            return "Account Maintenance"
        else:
            return "Other Debit"

    # If both columns are NaN or 0
    else:
        return "Uncategorized"
```

```python
df["Category"] = df.apply(lambda x: classify_transaction(x["DESCRIPTION"], x["DEPOSITS"],x["WITHDRAWALS"]), axis=1)
```
